```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Cargar las imágenes
ruta1 = "Imagenes/1050.png"
img1 = cv2.imread(ruta1)
img_rgb1 = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
ruta2 = "Imagenes/1051.png"
img2 = cv2.imread(ruta2)
img_rgb2 = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
ruta3 = "Imagenes/1052.png"
img3 = cv2.imread(ruta3)
img_eg3 = cv2.cvtColor(img3, cv2.COLOR_BGR2GRAY)
ruta4 = "Imagenes/1053.png"
img4 = cv2.imread(ruta4)
img_eg4 = cv2.cvtColor(img4, cv2.COLOR_BGR2GRAY)

# Mostrar una imagen de prueba
print("Imagene de prueba")
plt.figure(figsize=(5, 5)) # Creacion de la ventana
plt.imshow(img_rgb1) # Mostrar la imagen en la ventana
plt.title("Imagen 1") # Ponerle un titulo a la imagen
plt.axis('off') # NO mostrar los ejes (para que se vea limpia)
plt.tight_layout()
plt.show() # Mostrar la ventana

# IMAGEN ORIGINAL -------------------------------------------------------------------------------------------------------------------------------------
print("\nImagenes Originales")

plt.figure(figsize=(20, 5)) 

plt.subplot(1,4,1)
plt.imshow(img_rgb1)
plt.title("Imagen 1 Original RGB", fontsize=18) 
plt.axis('off') 

plt.subplot(1,4,2)
plt.imshow(img_rgb2)
plt.title("Imagen 2 Original RGB", fontsize=18) 
plt.axis('off') 

plt.subplot(1,4,3)
plt.imshow(img_eg3, cmap='gray')
plt.title("Imagen 3 Original Escala Grises", fontsize=18) 
plt.axis('off') 

plt.subplot(1,4,4)
plt.imshow(img_eg4, cmap='gray')
plt.title("Imagen 4 Original Escala Grises", fontsize=18) 
plt.axis('off') 

plt.tight_layout()
plt.show() 


# TÉCNICA: TRANSFORMACIÓN DE INTENSIDAD ---------------------------------------------------------------------------------------------------------------
# Se quiere aumentar el contraste en un 20% y el brillo en 30 unidades
print("\nImagenes transformando la intensidad")
contraste = 1.5 # 1.0 contraste original
brillo = 50 # 0 brillo original
img_transformacion_intensidad1 = cv2.convertScaleAbs(img_rgb1, alpha=contraste, beta=brillo)
img_transformacion_intensidad2 = cv2.convertScaleAbs(img_rgb2, alpha=contraste, beta=brillo)
img_transformacion_intensidad3 = cv2.convertScaleAbs(img_eg3, alpha=contraste, beta=brillo)
img_transformacion_intensidad4 = cv2.convertScaleAbs(img_eg4, alpha=contraste, beta=brillo)

plt.figure(figsize=(20, 5)) 

plt.subplot(1,4,1)
plt.imshow(img_transformacion_intensidad1)
plt.title("Imagen 1 - Transformación Intensidad", fontsize=18) 
plt.axis('off') 

plt.subplot(1,4,2)
plt.imshow(img_transformacion_intensidad2)
plt.title("Imagen 2 - Transformación Intensidad", fontsize=18) 
plt.axis('off') 

plt.subplot(1,4,3)
plt.imshow(img_transformacion_intensidad3, cmap='gray')
plt.title("Imagen 3 - Transformación Intensidad", fontsize=18) 
plt.axis('off') 

plt.subplot(1,4,4)
plt.imshow(img_transformacion_intensidad4, cmap='gray')
plt.title("Imagen 4 - Transformación Intensidad", fontsize=18) 
plt.axis('off') 

plt.tight_layout()
plt.show() 

# TÉCNICA: PROCESAMIENTO DEL HISTOGRAMA ---------------------------------------------------------------------------------------------------------------
print("\nImagenes Procesamiento de histograma")
# Convertir las imagenes a escala de grises (Necesario para ecualizar)
img_hist1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
img_hist2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)
img_hist3 = cv2.cvtColor(img3, cv2.COLOR_BGR2GRAY)
img_hist4 = cv2.cvtColor(img4, cv2.COLOR_BGR2GRAY)

# Se aplica la ecualización de histograma
img_ecualizada1 = cv2.equalizeHist(img_hist1)
img_ecualizada2 = cv2.equalizeHist(img_hist2)
img_ecualizada3 = cv2.equalizeHist(img_hist3)
img_ecualizada4 = cv2.equalizeHist(img_hist4)

# Se pasan a RGB (1 y 2)
img_ecualizada1 = cv2.cvtColor(img_ecualizada1, cv2.COLOR_BGR2RGB)
img_ecualizada2 = cv2.cvtColor(img_ecualizada2, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(20, 5)) 

plt.subplot(1,4,1)
plt.imshow(img_ecualizada1)
plt.title("Imagen 1 - P. histograma (ecualización)", fontsize=18) 
plt.axis('off') 

plt.subplot(1,4,2)
plt.imshow(img_ecualizada2)
plt.title("Imagen 2 - P. histograma (ecualización)", fontsize=18) 
plt.axis('off') 

plt.subplot(1,4,3)
plt.imshow(img_ecualizada3, cmap='gray')
plt.title("Imagen 3 - P. histograma (ecualización)", fontsize=18) 
plt.axis('off') 

plt.subplot(1,4,4)
plt.imshow(img_ecualizada4, cmap='gray')
plt.title("Imagen 4 - P. histograma (ecualización)", fontsize=18) 
plt.axis('off') 

plt.tight_layout()
plt.show() 


# TÉCNICA: OPERADORES ARITMÉTICOS ---------------------------------------------------------------------------------------------------------------------
print("\nImagenes Operadores Aritméticos:\n\n") 
# Sacar la matriz de la imagen
num_a_operar = 50 # 50 es lo que se le va a operar a cada número de pixel
img_matriz1 = np.ones(img1.shape, dtype="uint8") * num_a_operar 
img_matriz2 = np.ones(img2.shape, dtype="uint8") * num_a_operar
img_matriz3 = np.ones(img3.shape, dtype="uint8") * num_a_operar
img_matriz4 = np.ones(img4.shape, dtype="uint8") * num_a_operar

# Suma
img_suma1 = cv2.add(img1, img_matriz1)
img_suma2 = cv2.add(img2, img_matriz2)
img_suma3 = cv2.add(img3, img_matriz3)
img_suma4 = cv2.add(img4, img_matriz4)
# a RGB 
img_rgb_suma1 = cv2.cvtColor(img_suma1, cv2.COLOR_BGR2RGB)
img_rgb_suma2 = cv2.cvtColor(img_suma2, cv2.COLOR_BGR2RGB)
img_rgb_suma3 = cv2.cvtColor(img_suma3, cv2.COLOR_BGR2GRAY)
img_rgb_suma4 = cv2.cvtColor(img_suma4, cv2.COLOR_BGR2GRAY)

# Resta
img_resta1 = cv2.subtract(img1, img_matriz1)
img_resta2 = cv2.subtract(img2, img_matriz2)
img_resta3 = cv2.subtract(img3, img_matriz3)
img_resta4 = cv2.subtract(img4, img_matriz4)
# a RGB 
img_rgb_resta1 = cv2.cvtColor(img_resta1, cv2.COLOR_BGR2RGB)
img_rgb_resta2 = cv2.cvtColor(img_resta2, cv2.COLOR_BGR2RGB)
img_rgb_resta3 = cv2.cvtColor(img_resta3, cv2.COLOR_BGR2GRAY)
img_rgb_resta4 = cv2.cvtColor(img_resta4, cv2.COLOR_BGR2GRAY)

# Multiplicacion
img_mult1 = cv2.multiply(img1, img_matriz1)
img_mult2 = cv2.multiply(img2, img_matriz2)
img_mult3 = cv2.multiply(img3, img_matriz3)
img_mult4 = cv2.multiply(img4, img_matriz4)
# a RGB 
img_rgb_mult1 = cv2.cvtColor(img_mult1, cv2.COLOR_BGR2RGB)
img_rgb_mult2 = cv2.cvtColor(img_mult2, cv2.COLOR_BGR2RGB)
img_rgb_mult3 = cv2.cvtColor(img_mult3, cv2.COLOR_BGR2GRAY)
img_rgb_mult4 = cv2.cvtColor(img_mult4, cv2.COLOR_BGR2GRAY)

# Division
img_div1 = cv2.divide(img1, img_matriz1)
img_div2 = cv2.divide(img2, img_matriz2)
img_div3 = cv2.divide(img3, img_matriz3)
img_div4 = cv2.divide(img4, img_matriz4)
# a RGB 
img_rgb_div1 = cv2.cvtColor(img_div1, cv2.COLOR_BGR2RGB)
img_rgb_div2 = cv2.cvtColor(img_div2, cv2.COLOR_BGR2RGB)
img_rgb_div3 = cv2.cvtColor(img_div3, cv2.COLOR_BGR2GRAY)
img_rgb_div4 = cv2.cvtColor(img_div4, cv2.COLOR_BGR2GRAY)

#Mostrar SUMA
print("Operaciones aritméticas: SUMA")
plt.figure(figsize=(20, 5)) 

plt.subplot(1,4,1)
plt.imshow(img_rgb_suma1)
plt.title("Imagen 1 - O. Artiméticas: Suma") 
plt.axis('off') 

plt.subplot(1,4,2)
plt.imshow(img_rgb_suma2)
plt.title("Imagen 2 - O. Artiméticas: Suma") 
plt.axis('off') 

plt.subplot(1,4,3)
plt.imshow(img_rgb_suma3, cmap='gray')
plt.title("Imagen 3 - O. Artiméticas: Suma") 
plt.axis('off') 

plt.subplot(1,4,4)
plt.imshow(img_rgb_suma4, cmap='gray')
plt.title("Imagen 4 - O. Artiméticas: Suma") 
plt.axis('off') 

plt.tight_layout()
plt.show() 

#Mostrar RESTA
print("Operaciones aritméticas: RESTA")
plt.figure(figsize=(20, 5)) 

plt.subplot(1,4,1)
plt.imshow(img_rgb_resta1)
plt.title("Imagen 1 - O. Artiméticas: Resta") 
plt.axis('off') 

plt.subplot(1,4,2)
plt.imshow(img_rgb_resta2)
plt.title("Imagen 2 - O. Artiméticas: Resta") 
plt.axis('off') 

plt.subplot(1,4,3)
plt.imshow(img_rgb_resta3, cmap='gray')
plt.title("Imagen 3 - O. Artiméticas: Resta") 
plt.axis('off') 

plt.subplot(1,4,4)
plt.imshow(img_rgb_resta4, cmap='gray')
plt.title("Imagen 4 - O. Artiméticas: Resta") 
plt.axis('off') 

plt.tight_layout()
plt.show() 

#Mostrar MULTIPLICACION
print("Operaciones aritméticas: MULTIPLICACION")
plt.figure(figsize=(20, 5)) 

plt.subplot(1,4,1)
plt.imshow(img_rgb_mult1)
plt.title("Imagen 1 - O. Artiméticas: Multiplicación", fontsize=16) 
plt.axis('off') 

plt.subplot(1,4,2)
plt.imshow(img_rgb_mult2)
plt.title("Imagen 2 - O. Artiméticas: Multiplicación", fontsize=16) 
plt.axis('off') 

plt.subplot(1,4,3)
plt.imshow(img_rgb_mult3, cmap='gray')
plt.title("Imagen 3 - O. Artiméticas: Multiplicación", fontsize=16) 
plt.axis('off') 

plt.subplot(1,4,4)
plt.imshow(img_rgb_mult4, cmap='gray')
plt.title("Imagen 4 - O. Artiméticas: Multiplicación", fontsize=16) 
plt.axis('off') 

plt.tight_layout()
plt.show() 

#Mostrar DIVISION
print("Operaciones aritméticas: DIVISION")
plt.figure(figsize=(20, 5)) 

plt.subplot(1,4,1)
plt.imshow(img_rgb_div1)
plt.title("Imagen 1 - O. Artiméticas: División", fontsize=16) 
plt.axis('off') 

plt.subplot(1,4,2)
plt.imshow(img_rgb_div2)
plt.title("Imagen 2 - O. Artiméticas: División", fontsize=16) 
plt.axis('off') 

plt.subplot(1,4,3)
plt.imshow(img_rgb_div3, cmap='gray')
plt.title("Imagen 3 - O. Artiméticas: División", fontsize=16) 
plt.axis('off') 

plt.subplot(1,4,4)
plt.imshow(img_rgb_div4, cmap='gray')
plt.title("Imagen 4 - O. Artiméticas: División", fontsize=16) 
plt.axis('off') 

plt.tight_layout()
plt.show() 



# TÉCNICA: CORRECCIÓN GAMMA (LEY DE POTENCIA)-------------------------------------------------------------------------------------------------------
print("\nImagenes Corrección Gamma:\n\n") 
gamma = 1.5
inv_gamma = 1.0 / gamma
table = np.array([(i / 255.0) ** inv_gamma * 255 for i in range(256)]).astype("uint8")

img_gamma1 = cv2.LUT(img_rgb1, table)
img_gamma2 = cv2.LUT(img_rgb2, table)
img_gamma3 = cv2.LUT(img_eg3, table)
img_gamma4 = cv2.LUT(img_eg4, table)

plt.figure(figsize=(20, 5))

plt.subplot(1,4,1)
plt.imshow(img_gamma1)
plt.title("Imagen 1 - Corrección Gamma", fontsize=16)
plt.axis('off')

plt.subplot(1,4,2)
plt.imshow(img_gamma2)
plt.title("Imagen 2 - Corrección Gamma", fontsize=16)
plt.axis('off')

plt.subplot(1,4,3)
plt.imshow(img_gamma3, cmap='gray')
plt.title("Imagen 3 - Corrección Gamma", fontsize=16)
plt.axis('off')

plt.subplot(1,4,4)
plt.imshow(img_gamma4, cmap='gray')
plt.title("Imagen 4 - Corrección Gamma", fontsize=16)
plt.axis('off')

plt.tight_layout()
plt.show()


# TÉCNICA: TRANSFORMACIÓN LOGARITMICA-----------------------------------------------------------------------------------------------------------------
print("\nImagenes con Transformación Logarítmica")
def log_transform(image):
    c = 255 / np.log(1 + np.max(image))
    log_img = c * (np.log(1 + image.astype(np.float32)))
    return np.array(log_img, dtype=np.uint8)

img_log1 = log_transform(img_rgb1)
img_log2 = log_transform(img_rgb2)
img_log3 = log_transform(img_eg3)
img_log4 = log_transform(img_eg4)

plt.figure(figsize=(20, 5))

plt.subplot(1,4,1)
plt.imshow(img_log1)
plt.title("Imagen 1 - Logarítmica", fontsize=16)
plt.axis('off')

plt.subplot(1,4,2)
plt.imshow(img_log2)
plt.title("Imagen 2 - Logarítmica", fontsize=16)
plt.axis('off')

plt.subplot(1,4,3)
plt.imshow(img_log3, cmap='gray')
plt.title("Imagen 3 - Logarítmica", fontsize=16)
plt.axis('off')

plt.subplot(1,4,4)
plt.imshow(img_log4, cmap='gray')
plt.title("Imagen 4 - Logarítmica", fontsize=16)
plt.axis('off')

plt.tight_layout()
plt.show()


# TÉCNICA: NEGATIVO------------------------------------------------------------------------------------------------------------------------------------
print("\nImagenes con Efecto Negativo")
img_neg1 = 255 - img_rgb1
img_neg2 = 255 - img_rgb2
img_neg3 = 255 - img_eg3
img_neg4 = 255 - img_eg4

plt.figure(figsize=(20, 5))

plt.subplot(1,4,1)
plt.imshow(img_neg1)
plt.title("Imagen 1 - Negativo", fontsize=16)
plt.axis('off')

plt.subplot(1,4,2)
plt.imshow(img_neg2)
plt.title("Imagen 2 - Negativo", fontsize=16)
plt.axis('off')

plt.subplot(1,4,3)
plt.imshow(img_neg3, cmap='gray')
plt.title("Imagen 3 - Negativo", fontsize=16)
plt.axis('off')

plt.subplot(1,4,4)
plt.imshow(img_neg4, cmap='gray')
plt.title("Imagen 4 - Negativo", fontsize=16)
plt.axis('off')

plt.tight_layout()
plt.show()

# OTROS CÁLCULOS------------------------------------------------------------------------------------------------------------------------------------
i1 =  cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
i2 = img_gamma1
i3 = img_ecualizada1


plt.figure(figsize=(4, 4))
plt.imshow(i1, cmap='gray')
plt.title("Imagen 1 Original RGB", fontsize=10)
print("\nValores Imagen Original: ")
mediai1 = np.mean(i1)
medianai1 = np.median(i1)
varianzai1 = np.var(i1)
print("Media: ", round(mediai1, 2), "|Mediana: ", round(medianai1, 2), "|Varianza: ", round(varianzai1, 2))
plt.axis('off') 
plt.tight_layout()

plt.figure(figsize=(4, 4))
plt.imshow(i2)
plt.title("Imagen 1 Original Corrección Gamma", fontsize=10) 
print("\nValores Imagen con Corrección Gamma: ")
mediai2 = np.mean(i2)
medianai2 = np.median(i2)
varianzai2 = np.var(i2)
print("Media: ", round(mediai2, 2), "|Mediana: ", round(medianai2, 2), "|Varianza: ", round(varianzai2, 2))
plt.axis('off') 

print("\nValores Imagen con Ecualización de Histograma: ")
mediai3 = np.mean(i3)
medianai3 = np.median(i3)
varianzai3 = np.var(i3)
print("Media: ", round(mediai3, 2), "|Mediana: ", round(medianai3, 2), "|Varianza: ", round(varianzai3, 2))
plt.figure(figsize=(4, 4))
plt.imshow(i3)
plt.title("Imagen 1 Original Ecualización de Histograma", fontsize=10) 
plt.axis('off')
plt.show()
```
